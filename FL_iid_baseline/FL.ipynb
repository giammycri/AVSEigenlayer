```python
# Ora quello che si deve fare, visto che il dataset è stato cariacato correttamente, è creare il modello di rete da utilizzare.
# Quello che dovrebbe essere fatto è utilizzare i dati mock per vedere se il modello può andare bene con la rete che andremo ad utilizzare. Nel nostro caso, andremo a saltare questo passaggio poichè supponiamo che tutti i partecipanti utilizzino lo stesso dataset poichè l'obiettivo è valutare la robustezza del modello quando va a fare l'aggregazione dei pesi (FedAvg).


# Una volta creata la funzione, e una volta che abbiamo visto che essa funziona (su dati privati, ma alla fine secondo il nostro ragionamento, tutti avranno gli stessi dati), dobbiamo fare il wrapping della funzione per renderla disponibile a tutti gli altri.

@sy.syft_function()


#def modello (X_train,X_test,Y_train,Y_test):
def modello (X_train, Y_train):

    import tensorflow as tf
    from tensorflow.keras import layers, models

    # Normalizza i dati
    X_train = X_train.astype('float32') / 255
    #X_test = X_test.astype('float32') / 255

    # Aggiungi una dimensione per il canale di colore (grigio)
    X_train = X_train.reshape(X_train.shape[0], 28, 28, 1)
    #X_test = X_test.reshape(X_test.shape[0], 28, 28, 1)
    
    # Definisci il modello CNN
    model = models.Sequential([
        layers.Conv2D(16, (1, 1), activation='relu', input_shape=(28, 28, 1)),
        #layers.MaxPooling2D((2, 2)),
        #layers.Conv2D(16, (1, 1), activation='relu'),
        #layers.MaxPooling2D((2, 2)),
        #layers.Conv2D(64, (3, 3), activation='relu'),
        layers.Flatten(),
        #layers.Dense(64, activation='relu'),
        layers.Dense(10, activation='softmax')
    ])

    # Compila il modello
    model.compile(optimizer='adam',
                loss='sparse_categorical_crossentropy',
                metrics=['accuracy'])

    # Addestra il modello
    #model.fit(X_train, Y_train, epochs=5, batch_size=64, validation_data=(X_test, Y_test))
    model.fit(X_train, Y_train, epochs=1, batch_size=64)
    
    # Ottieni i pesi del modello
    pesi_del_modello = model.get_weights()
    
    return model, pesi_del_modello
```

```python
# ============================================
# TRAINING LOCALE SUI CLIENTS
# ============================================
import os
import time

print("🏋️ Inizio training locale sui clients...\n")

modello_alice_list = []
modello_pesi_alice_list = []

# MISURA TEMPO TRAINING
training_start_time = time.time()

# Esegui tutti gli script nella cartella 'scripts'
scripts_dir = 'scripts'
for script_file in os.listdir(scripts_dir):
    if script_file.endswith('.py'):
        script_path = os.path.join(scripts_dir, script_file)
        
        # Ottieni l'indice dell'utente dall'indice dello script
        user_index = int(script_file.split('_')[1].split('.')[0])
        
        # Esegui la funzione user_function per l'utente corrente
        data_prefix = f"_a_{user_index}"
        modello_alice, modello_pesi_alice = user_function(
            X_train=eval(f"train_X{data_prefix}"), 
            Y_train=eval(f"train_y{data_prefix}")
        )
        
        # Aggiungi i risultati alla lista
        modello_alice_list.append(modello_alice)
        modello_pesi_alice_list.append(modello_pesi_alice)

training_end_time = time.time()
training_total_time = training_end_time - training_start_time

print(f"\n✅ Training completato!")
print(f"⏱️  Tempo totale training: {training_total_time:.2f}s")
print(f"⏱️  Tempo medio per client: {training_total_time/len(modello_alice_list):.2f}s")
print(f"👥 Clients addestrati: {len(modello_alice_list)}\n")
```

```python
# Deposita i risultati
deposito_risultati = request.accept_by_depositing_result(modello_pesi_alice_list)

# ============================================
# VALIDAZIONE AVS + RACCOLTA METRICHE
# ============================================
import subprocess
import json
import os
import hashlib
import pickle
import time
import re

print("🔍 Validazione pesi con AVS EigenLayer...\n")

# ============================================
# RACCOLTA METRICHE SPERIMENTALI
# ============================================
experimental_results = {
    'num_clients': len(modello_pesi_alice_list),
    'training_time_total': training_total_time,
    'training_time_per_client': training_total_time / len(modello_pesi_alice_list),
    'clients': [],
    'validation_time_total': 0,
    'avg_weights_size_kb': 0,
    'avg_payload_size_bytes': 0,
    'total_gas_used': 0,
    'total_cost_eth': 0
}

validated_models = []

# MISURA TEMPO TOTALE VALIDAZIONE
validation_start = time.time()

for i in range(len(modello_pesi_alice_list)):
    weights = modello_pesi_alice_list[i]
    model = modello_alice_list[i]
    
    print(f"📤 Validazione client {i}...")
    
    # Misura tempo per questo client
    client_validation_start = time.time()
    
    try:
        # Calcola hash dei pesi
        weights_bytes = pickle.dumps(weights)
        weights_size = len(weights_bytes)
        weights_hash = hashlib.sha256(weights_bytes).digest()
        
        # Calcola dimensione del payload (hash + clientId + claimedResult)
        payload_size = 32 + 32 + 32  # 96 bytes totali
        
        print(f"   Dimensione pesi: {weights_size:,} bytes ({weights_size/1024:.2f} KB)")
        print(f"   Hash: {weights_hash.hex()[:16]}...")
        print(f"   Payload size: {payload_size} bytes")
        
        # Avvia il comando con la signature corretta
        process = subprocess.Popen(
            [
                "devkit", "avs", "call",
                "signature=(bytes32,uint256,uint256)",
                f"args=(0x{weights_hash.hex()},{i},1)"
            ],
            cwd=os.path.expanduser("~/devkit-cli/my-avs-project"),
            stdout=subprocess.PIPE,
            stderr=subprocess.PIPE,
            text=True
        )
        
        # Aspetta che finisca (max 120s)
        try:
            stdout, stderr = process.communicate(timeout=120)
        except subprocess.TimeoutExpired:
            process.kill()
            print(f"   ❌ Timeout")
            client_metrics = {
                'client_id': i,
                'valid': False,
                'validation_time': time.time() - client_validation_start,
                'weights_size_bytes': weights_size,
                'weights_size_kb': weights_size / 1024,
                'payload_size_bytes': payload_size,
                'error': 'Timeout'
            }
            experimental_results['clients'].append(client_metrics)
            print()
            continue
        
        # Aspetta che il JSON venga scritto nei log
        print(f"   ⏳ Aspetto che il task venga processato...")
        time.sleep(5)
        
        # Calcola tempo totale
        client_validation_time = time.time() - client_validation_start
        
        # Leggi l'output completo
        all_output = stderr + "\n" + stdout
        
        # ESTRAI TX HASH - cerca la riga con "✅  [Success] Hash:" 
        # (transaction hash della creazione del task on-chain)
        tx_hash = None
        # Cerca nella parte iniziale dell'output prima di "Task created successfully"
        tx_section = all_output.split("Task created successfully")[0]
        tx_hash_match = re.search(r'Hash:\s*(0x[a-fA-F0-9]{64})', tx_section)
        if tx_hash_match:
            tx_hash = tx_hash_match.group(1)
        
        # ESTRAI TASK HASH - cerca "Task hash: 0x..." o "- Task hash: 0x..."
        task_hash = None
        task_hash_match = re.search(r'-?\s*Task hash:\s*(0x[a-fA-F0-9]{64})', all_output)
        if task_hash_match:
            task_hash = task_hash_match.group(1)
        
        # ESTRAI GAS USED dalla riga "Paid: X.XXX ETH (YYYYY gas * ...)"
        gas_used = None
        gas_match = re.search(r'Paid:\s*[\d.]+\s*ETH\s*\((\d+)\s*gas', all_output)
        if gas_match:
            gas_used = int(gas_match.group(1))
        
        # ESTRAI GAS SCRIPT dalla riga "Estimated total gas used for script: XXXXX"
        gas_script = None
        gas_script_match = re.search(r'Estimated total gas used for script:\s*(\d+)', all_output)
        if gas_script_match:
            gas_script = int(gas_script_match.group(1))
        
        # ESTRAI COST ETH dalla riga "Paid: X.XXXXXX ETH"
        cost_eth = None
        cost_match = re.search(r'Paid:\s*([\d.]+)\s*ETH', all_output)
        if cost_match:
            cost_eth = float(cost_match.group(1))
        
        # ESTRAI GAS PRICE dalla riga con "X.XXX gwei"
        gas_price_gwei = None
        gas_price_match = re.search(r'(\d+\.?\d*)\s*gwei', all_output)
        if gas_price_match:
            gas_price_gwei = float(gas_price_match.group(1))
        elif gas_used and cost_eth:
            # Calcola gas price da cost_eth / gas_used
            gas_price_wei = (cost_eth * 1e18) / gas_used
            gas_price_gwei = gas_price_wei / 1e9
        
        # Cerca il JSON finale per lo status
        json_match = re.search(r'\{"taskHash"[^}]+\}', all_output)
        is_valid = False
        
        if json_match:
            json_str = json_match.group(0)
            try:
                result = json.loads(json_str)
                if result.get("status") == "verified":
                    is_valid = True
            except:
                pass
        
        # Fallback: cerca "Task execution completed"
        if not is_valid and "Task execution completed successfully" in all_output:
            is_valid = True
        
        if is_valid:
            validated_models.append(model)
            print(f"   ✅ VALIDATO")
            print(f"   ⏱️  Tempo validazione: {client_validation_time:.2f}s")
            if tx_hash:
                print(f"   🔗 TX Hash: {tx_hash[:18]}...")
            if task_hash:
                print(f"   📝 Task Hash: {task_hash[:18]}...")
            if gas_used:
                print(f"   ⛽ Gas transazione: {gas_used:,}")
            if gas_script:
                print(f"   ⛽ Gas script totale: {gas_script:,}")
            if gas_price_gwei:
                print(f"   💵 Gas price: {gas_price_gwei:.9f} gwei")
            if cost_eth:
                print(f"   💰 Costo: {cost_eth:.15f} ETH")
            
            client_metrics = {
                'client_id': i,
                'valid': True,
                'validation_time': client_validation_time,
                'weights_size_bytes': weights_size,
                'weights_size_kb': weights_size / 1024,
                'payload_size_bytes': payload_size,
                'weights_hash': weights_hash.hex(),
                'tx_hash': tx_hash,
                'task_hash': task_hash,
                'gas_used': gas_used,
                'gas_script': gas_script,
                'gas_price_gwei': gas_price_gwei,
                'cost_eth': cost_eth
            }
            
            # Aggiungi ai totali (usa il gas della transazione effettiva)
            if gas_used:
                experimental_results['total_gas_used'] += gas_used
            if cost_eth:
                experimental_results['total_cost_eth'] += cost_eth
        else:
            print(f"   ❌ RIFIUTATO")
            print(f"   ⏱️  Tempo: {client_validation_time:.2f}s")
            
            client_metrics = {
                'client_id': i,
                'valid': False,
                'validation_time': client_validation_time,
                'weights_size_bytes': weights_size,
                'weights_size_kb': weights_size / 1024,
                'payload_size_bytes': payload_size,
                'tx_hash': tx_hash,
                'task_hash': task_hash,
                'gas_used': gas_used,
                'gas_script': gas_script,
                'gas_price_gwei': gas_price_gwei,
                'cost_eth': cost_eth
            }
        
        experimental_results['clients'].append(client_metrics)
        
    except Exception as e:
        print(f"   ❌ Errore: {e}")
        client_validation_time = time.time() - client_validation_start
        experimental_results['clients'].append({
            'client_id': i,
            'valid': False,
            'validation_time': client_validation_time,
            'weights_size_bytes': weights_size if 'weights_size' in locals() else 0,
            'weights_size_kb': (weights_size / 1024) if 'weights_size' in locals() else 0,
            'payload_size_bytes': 96,
            'error': str(e)
        })
    
    print()

validation_end = time.time()
experimental_results['validation_time_total'] = validation_end - validation_start
experimental_results['validation_time_per_client'] = experimental_results['validation_time_total'] / len(modello_pesi_alice_list)

# Calcola medie
total_weights_size = sum(c['weights_size_bytes'] for c in experimental_results['clients'])
experimental_results['avg_weights_size_kb'] = total_weights_size / len(experimental_results['clients']) / 1024

total_payload_size = sum(c.get('payload_size_bytes', 96) for c in experimental_results['clients'])
experimental_results['avg_payload_size_bytes'] = total_payload_size / len(experimental_results['clients'])

# Calcola costi medi (solo per task validati)
num_valid = len([c for c in experimental_results['clients'] if c.get('valid', False)])
if num_valid > 0 and experimental_results['total_gas_used'] > 0:
    experimental_results['avg_gas_used'] = experimental_results['total_gas_used'] / num_valid
if num_valid > 0 and experimental_results['total_cost_eth'] > 0:
    experimental_results['avg_cost_eth'] = experimental_results['total_cost_eth'] / num_valid

# ============================================
# STAMPA RIEPILOGO
# ============================================
print(f"\n{'='*70}")
print(f"📊 RIEPILOGO ESPERIMENTO - {experimental_results['num_clients']} CLIENTS")
print(f"{'='*70}")

print(f"\n🏋️ TRAINING:")
print(f"   Tempo totale: {experimental_results['training_time_total']:.2f}s")
print(f"   Tempo per client: {experimental_results['training_time_per_client']:.2f}s")

print(f"\n🔍 VALIDAZIONE AVS:")
print(f"   Clients validati: {len(validated_models)}/{experimental_results['num_clients']}")
print(f"   Tempo totale: {experimental_results['validation_time_total']:.2f}s")
print(f"   Tempo per client: {experimental_results['validation_time_per_client']:.2f}s")

print(f"\n💰 COSTI BLOCKCHAIN:")
if experimental_results['total_gas_used'] > 0:
    print(f"   Gas totale: {experimental_results['total_gas_used']:,}")
    print(f"   Gas medio: {experimental_results.get('avg_gas_used', 0):,.0f}")
if experimental_results['total_cost_eth'] > 0:
    print(f"   Costo totale: {experimental_results['total_cost_eth']:.15f} ETH")
    print(f"   Costo medio: {experimental_results.get('avg_cost_eth', 0):.15f} ETH")

print(f"\n📦 DIMENSIONI:")
print(f"   Dim. media pesi: {experimental_results['avg_weights_size_kb']:.2f} KB")
print(f"   Dim. media payload: {experimental_results['avg_payload_size_bytes']:.0f} bytes")

print(f"{'='*70}\n")

# ============================================
# FEDERATED AVERAGING
# ============================================
from federated_averaging import federated_averaging

if len(validated_models) == 0:
    print("⚠️ Nessun peso validato! Usando tutti i modelli per test...")
    validated_models = modello_alice_list

print("🔄 Esecuzione Federated Averaging...")
fedavg_start = time.time()
global_model = federated_averaging(modello_test, validated_models)
fedavg_time = time.time() - fedavg_start
experimental_results['fedavg_time'] = fedavg_time
print(f"✅ FedAvg completato in {fedavg_time:.2f}s!\n")

# ============================================
# SALVA RISULTATI
# ============================================
results_file = f'experimental_results_{experimental_results["num_clients"]}_clients.json'
with open(results_file, 'w') as f:
    json.dump(experimental_results, f, indent=2)

print(f"💾 Risultati salvati in: {results_file}\n")
```
